##### Valid Anagram
Given two strings s and t, return true if the two strings are anagrams of each other, otherwise return false.\
Two strings are anagrams if they contain the same characters, with each character appearing the same number of times, regardless of order.

##### Test Cases
Example 1:\
Input: s = "racecar", t = "carrace"\
Output: true

Example 2:\
Input: s = "jar", t = "jam"\
Output: false

```python
from typing import *
```

```python
# SOLUTION 1
class Solution:
    def isAnagram(self, s: str, t: str) -> bool:
        if len(s) != len(t):
            return False
        countS, countT = {}, {}
        for i in range(len(s)):
            # count[char] = (existing_count if exists or 0) + 1
            countS[s[i]] = 1 + countS.get(s[i], 0)
            countT[t[i]] = 1 + countT.get(t[i], 0)
        return countS == countT

sol = Solution()
print(sol.isAnagram("racecar", "carrace"))
```

```python
# SOLUTION 2
class Solution:
    def isAnagram(self, s: str, t: str) -> bool:
        if len(s) != len(t):
            return False
        for i in set(s):
            if s.count(i) != t.count(i):
                return False
        return True
    
sol = Solution()
print(sol.isAnagram("racecar", "carrace"))
```

##### Contains Duplicate
Given an integer array nums, return true if any value appears more than once in the array, otherwise return false.

##### Test Cases
Example 1:\
Input: nums = [1, 2, 3, 3]\
Output: true

Example 2:\
Input: nums = [1, 2, 3, 4]\
Output: false

```python
from typing import *
```

```python
# SOLUTION: Intermediate
class Solution:
    def hasDuplicate(self, nums: List[int]) -> bool:
        nums_set = set()
        for n in nums: #Time Complexity: O(n)
            if n in nums_set:
                return True
            nums_set.add(n)  
        return False

sol = Solution()
print(sol.hasDuplicate([1, 2, 3, 3]))
```

```python
# SOLUTION: Optimized
class Solution:
    def hasDuplicate(self, nums: List[int]) -> bool:
        # Dictionary consumes less space than Sets
        # Reduced Space Complexity
        dict = {}
        for n in nums:
            if n in dict:
                return True
            dict[n] = 1
        return False
    
sol = Solution()
print(sol.hasDuplicate([1, 2, 3, 3]))
```

##### Two Sum
Given an array of integers `nums` and an integer `target`, return the indices `i` and `j` such that `nums[i] + nums[j] == target` and `i != j`.\
You may assume that *every* input has exactly one pair of indices `i` and `j` that satisfy the condition.\
Return the answer with the smaller index first.

##### Test Cases
Example 1:\
Input: \  
nums = [3,4,5,6], target = 7\
Output: [0,1]

Explanation: `nums[0] + nums[1] == 7`, so we return `[0, 1]`.

Example 2:\
Input: nums = [4,5,6], target = 10\
Output: [0,2]

Example 3:\
Input: nums = [5,5], target = 10\
Output: [0,1]

```python
from typing import *
```

```python
# SOLUTION: BRUTE FORCE 1
class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        a = 0
        b = 1
        while a <= len(nums):
            if nums[a] + nums[b] == target:
                return [a,b]
            b += 1
            if b == len(nums):
                a += 1
                b = a + 1

sol = Solution()
print(sol.twoSum([3,4,5,6], 7))
```

```python
# SOLUTION: BRUTE FORCE 2
class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        for i in range(len(nums)):
            diff = target - nums[i]
            if diff in nums:
                if nums.index(diff) > i:
                    return [i, nums.index(diff)]
                elif nums.index(diff) < i:
                    return [nums.index(diff), i]
                
sol = Solution()
print(sol.twoSum([3,4,5,6], 7))
```

```python
# SOLUTION: INTERMEDIATE
class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        temp = []
        for i in range(len(nums)):
            diff = target - nums[i]
            if diff in temp:
                return [temp.index(diff), i]
            temp.append(nums[i])
                            
sol = Solution()
print(sol.twoSum([3,4,5,6], 7))
```

```python
# SOLUTION: OPTIMIZED 1
class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        dict = {}
        for i in range(len(nums)):
            diff = target - nums[i]
            if diff in dict:
                return [dict[diff], i]
            dict[nums[i]] = i
                                        
sol = Solution()
print(sol.twoSum([3,4,5,6], 7))
```

```python
# SOLUTION: OPTIMIZED 2
class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        dict = {}
        for i, n in enumerate(nums):
            diff = target - n
            if diff in dict:
                return [dict[diff], i]
            dict[n] = i
                                            
sol = Solution()
print(sol.twoSum([3,4,5,6], 7))
```

##### Merge Two Sorted Linked Lists
You are given the heads of two sorted linked lists list1 and list2.\
Merge the two lists into one sorted linked list and return the head of the new sorted linked list.\
The new list should be made up of nodes from list1 and list2.

##### Test Cases
Example 1:\
Input: list1 = [1,2,4], list2 = [1,3,5]\
Output: [1,1,2,3,4,5]

Example 2:\
Input: list1 = [], list2 = [1,2]\
Output: [1,2]

Example 3:\
Input: list1 = [], list2 = []\
Output: []

```python
from typing import *
```

```python
# SOLUTION: BRUTE FORCE

# Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

class Solution:
    def mergeTwoLists(self, list1: Optional[ListNode], list2: Optional[ListNode]) -> Optional[ListNode]:
        l1, l2 = [] , []
        t1 = list1
        while t1 is not None:
            l1.append(t1.val)
            t1 = t1.next

        t2 = list2   
        while t2 is not None:
            l2.append(t2.val)
            t2 = t2.next

        final = l1 + l2
        final = sorted(final)
        head = None
        for i in final:
            if head is None:
                head = ListNode(i)
            else:
                temp = ListNode(i)
                curr = head
                while curr.next is not None:
                    curr = curr.next
                curr.next = temp
        return head
```

```python
# SOLUTION: OPTIMIZED

# Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

class Solution:
    def mergeTwoLists(self, list1: Optional[ListNode], list2: Optional[ListNode]) -> Optional[ListNode]:
        d = ListNode() #To store the location of heas
        curr = d # For iteration of the new Linked List
        while list1 is not None and list2 is not None:
            if list1.val < list2.val: 
                curr.next = list1 #Set the next Node with smaller value
                list1 = list1.next # Forward the pointer to next node
            else:
                curr.next = list2
                list2 = list2.next
            curr = curr.next #Forward new pointer to next node
        if list1 is not None: # If there is still elements if list1 Node
            curr.next = list1 # Add the leftover to the new Linked List
        else:                 
            curr.next = list2

        return d.next # The first node is empty, so remove it
            
```

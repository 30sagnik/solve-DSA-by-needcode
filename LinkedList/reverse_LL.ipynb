##### Reverse Linked List
Given the beginning of a singly linked list head, reverse the list, and return the new beginning of the list.

##### Test Cases
Example 1:\
Input: head = [0,1,2,3]\
Output: [3,2,1,0]

Example 2:\
Input: head = []\
Output: []

```python
from typing import *
```

```python
# SOLUTION: OPTIMIZED

# Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

class Solution:
    def reverseList(self, head: Optional[ListNode]) -> Optional[ListNode]:
        prev, curr = None, head
        while curr is not None:
            temp = curr.next #To save the location to the next node
            curr.next = prev # Connect with the previous Node
            prev = curr 
            curr = temp # Use the saved location as curr.next will point to prev
        return prev
```
